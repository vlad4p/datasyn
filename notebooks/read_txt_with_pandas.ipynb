# Read `.txt` files with pandas

Examples aligned with Datacyber skills (`ingest-pandas-normalize`): comma- or **semicolon-separated** text, optional encoding.

Set `DATA_PATH` to a file your kernel can read (host path or a path inside Docker if the notebook runs there).

```python
from pathlib import Path

import pandas as pd

pd.__version__
```

## 1. Semicolon-separated values (common for `.txt` exports)

Use `sep=";"` and tune `encoding` if needed (`utf-8`, `latin-1`, etc.).

```python
# Example: replace with your file
DATA_PATH = Path("../data-local/EPH_usu_3_Trim_2025_txt/usu_individual_T325.txt")  # adjust
df_semi = pd.read_csv(DATA_PATH, sep=";", encoding="utf-8")
display(df_semi.head())
display(df_semi.info())
```

## 2. Comma-separated text in a `.txt` file

Default `read_csv` uses comma.

```python
DATA_PATH_COMMA = Path("../data-local/example_comma.txt")  # adjust

if DATA_PATH_COMMA.is_file():
    df_csv = pd.read_csv(DATA_PATH_COMMA, encoding="utf-8")
    display(df_csv.head())
else:
    print(f"Skip: file not found: {DATA_PATH_COMMA.resolve()}")
```

## 3. Probe only first rows (large files)

Use `nrows` to avoid loading everything into memory during exploration.

```python
if DATA_PATH.is_file():
    df_probe = pd.read_csv(DATA_PATH, sep=";", encoding="utf-8", nrows=5000)
    print(df_probe.shape)
    display(df_probe.head(10))
```

## 4. Optional: write normalized comma CSV (then DuckDB `ingest-csv`)

Same pattern as `skills/ingest-pandas-normalize/SKILL.md`.

```python
OUT_DIR = Path("../tmp")
OUT_DIR.mkdir(parents=True, exist_ok=True)

if DATA_PATH.is_file():
    df_out = pd.read_csv(DATA_PATH, sep=";", encoding="utf-8")
    out_path = OUT_DIR / f"{DATA_PATH.stem}.normalized.csv"
    df_out.to_csv(out_path, index=False)
    print("Wrote:", out_path.resolve())
else:
    print("No input file; nothing written.")
```
